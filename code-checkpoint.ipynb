# Forecast-Assisted Multivariate Time Series Analysis and Dashboard of Spain's Electricity Demand

This project uses Spain's hourly energy, price, forecast, and weather data to analyze and predict actual electricity demand.

The target variable is `total load actual`, because it represents the real electricity demand at each hour.

The dataset also includes `total load forecast`, which is an official external forecast signal. Therefore, this project is a **forecast-assisted electricity demand prediction** task, not a forecast-from-scratch task.

The main research question is:

**How can the official load forecast, time-series memory, weather, renewable forecast, and price-related features help predict actual hourly electricity demand in Spain?**

This section imports the required libraries for data cleaning, visualization, feature engineering, model training, and evaluation.

## 3. Data Cleaning and Merging

Before analysis, the time column must be converted into datetime format.

The energy data is at the national level, while the weather data is city-level. If we directly merge the two files, one timestamp may appear multiple times because there are five cities.

Therefore, we aggregate the numerical weather features by timestamp using the mean value. This creates a general weather signal for Spain and makes the weather data suitable for national-level forecast-assisted electricity demand prediction.


```python
# Convert time columns to datetime
energy["time"] = pd.to_datetime(energy["time"], utc=True)
weather["dt_iso"] = pd.to_datetime(weather["dt_iso"], utc=True)

# Remove timezone information for easier merging
energy["time"] = energy["time"].dt.tz_convert(None)
weather["dt_iso"] = weather["dt_iso"].dt.tz_convert(None)

# Rename weather time column
weather = weather.rename(columns={"dt_iso": "time"})

# Remove duplicate rows
energy = energy.drop_duplicates()
weather = weather.drop_duplicates()

print("Energy shape after removing duplicates:", energy.shape)
print("Weather shape after removing duplicates:", weather.shape)
```

```python
# Drop columns that are completely empty
energy = energy.dropna(axis=1, how="all")
weather = weather.dropna(axis=1, how="all")

print("Energy shape after dropping empty columns:", energy.shape)
print("Weather shape after dropping empty columns:", weather.shape)
```

```python
# Aggregate weather data by time using the mean of numerical columns
weather_numeric_cols = weather.select_dtypes(include=[np.number]).columns.tolist()
weather_avg = weather.groupby("time")[weather_numeric_cols].mean().reset_index()

print("Aggregated weather data shape:", weather_avg.shape)
display(weather_avg.head())
```

```python
# Merge energy and weather data by time
df = pd.merge(energy, weather_avg, on="time", how="inner")

# Sort data by time
df = df.sort_values("time").reset_index(drop=True)

print("Merged dataset shape:", df.shape)
display(df.head())
```

```python
# Handle abnormal pressure values before interpolation
if "pressure" in df.columns:
    pressure_anomaly = (df["pressure"] < 900) | (df["pressure"] > 1100)
    print("Number of abnormal pressure values:", pressure_anomaly.sum())

    # Replace abnormal pressure values with NaN
    df.loc[pressure_anomaly, "pressure"] = np.nan

# Handle missing values using time-based interpolation
df = df.set_index("time")

numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].interpolate(method="time").ffill().bfill()

df = df.reset_index()

print("Missing values after interpolation:")
display(df.isnull().sum().sort_values(ascending=False).head(15))

# Check pressure after cleaning
if "pressure" in df.columns:
    print("Pressure min after cleaning:", df["pressure"].min())
    print("Pressure max after cleaning:", df["pressure"].max())
```

```python
# Convert temperature from Kelvin to Celsius if needed
if "temp" in df.columns:
    if df["temp"].mean() > 100:
        df["temp_c"] = df["temp"] - 273.15
    else:
        df["temp_c"] = df["temp"]

display(df[["time", "total load actual", "temp_c"]].head())
```

## 5. Anomaly Detection and Treatment

Anomalies may disturb model training because they do not represent normal electricity demand patterns.

The IQR method is used to detect extreme values in `total load actual`.

To keep the hourly time structure, anomalous values are replaced by interpolation instead of deleting the whole row.

```python
target = "total load actual"

Q1 = df[target].quantile(0.25)
Q3 = df[target].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["is_anomaly"] = ((df[target] < lower_bound) | (df[target] > upper_bound)).astype(int)

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Number of detected anomalies:", df["is_anomaly"].sum())
```

```python
plt.figure(figsize=(14, 5))

plt.plot(df["time"], df[target], color=COLOR_DEMAND, linewidth=0.8, label="Original Demand")

plt.scatter(
    df.loc[df["is_anomaly"] == 1, "time"],
    df.loc[df["is_anomaly"] == 1, target],
    color=COLOR_ANOMALY,
    s=14,
    label="Detected Anomalies"
)

plt.title("Detected Anomalies in Electricity Demand")
plt.xlabel("Time")
plt.ylabel("Total Load Actual")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Replace anomalies with NaN and interpolate
df_clean = df.copy()
df_clean.loc[df_clean["is_anomaly"] == 1, target] = np.nan

df_clean = df_clean.set_index("time")
df_clean[target] = df_clean[target].interpolate(method="time").ffill().bfill()
df_clean = df_clean.reset_index()

plt.figure(figsize=(14, 5))

plt.plot(
    df["time"],
    df[target],
    color=COLOR_BEFORE,
    linewidth=0.7,
    label="Before Treatment"
)

plt.plot(
    df_clean["time"],
    df_clean[target],
    color=COLOR_DEMAND,
    linewidth=0.9,
    label="After Treatment"
)

plt.title("Electricity Demand Before and After Anomaly Treatment")
plt.xlabel("Time")
plt.ylabel("Total Load Actual")
plt.legend()
plt.tight_layout()
plt.show()
```

## 6. Feature Engineering

Feature engineering is important for time series demand prediction.

Electricity demand is not random. It is usually related to time patterns, previous demand values, weather, market context, and available forecast signals.

The following feature groups are used:

- Forecast signal: `total load forecast`
- Calendar features: `hour`, `dayofweek`, `month`, `is_weekend`
- Memory features: `lag_1`, `lag_24`, `lag_168`
- Rolling features: `rolling_mean_24`, `rolling_mean_168`
- Context features: weather, price, solar forecast, and wind forecast


```python
df_model = df_clean.copy()

# Time features
df_model["hour"] = df_model["time"].dt.hour
df_model["dayofweek"] = df_model["time"].dt.dayofweek
df_model["month"] = df_model["time"].dt.month
df_model["is_weekend"] = df_model["dayofweek"].isin([5, 6]).astype(int)

# Lag features
df_model["lag_1"] = df_model[target].shift(1)
df_model["lag_24"] = df_model[target].shift(24)
df_model["lag_168"] = df_model[target].shift(168)

# Rolling mean features
df_model["rolling_mean_24"] = df_model[target].shift(1).rolling(window=24).mean()
df_model["rolling_mean_168"] = df_model[target].shift(1).rolling(window=168).mean()

display(df_model[[
    "time", target, "hour", "dayofweek", "month", "is_weekend",
    "lag_1", "lag_24", "lag_168", "rolling_mean_24", "rolling_mean_168"
]].head(10))
```

## 7. Preprocessing

In this section, useful features are selected for model training.

For time series data, the dataset must be split chronologically.  
A random split is not suitable because it may allow future information to appear in the training set.

The first 80% of the data is used for training, and the last 20% is used for testing.

```python
# Candidate features
candidate_features = [
    # Time features
    "hour",
    "dayofweek",
    "month",
    "is_weekend",
    
    # Lag and rolling features
    "lag_1",
    "lag_24",
    "lag_168",
    "rolling_mean_24",
    "rolling_mean_168",
    
    # Weather features
    "temp_c",
    "humidity",
    "pressure",
    "wind_speed",
    "clouds_all",
    "rain_1h",
    "snow_3h",
    
    # Price and forecast-related features
    "total load forecast",
    "price day ahead",
    "forecast solar day ahead",
    "forecast wind onshore day ahead",
    "forecast wind offshore eday ahead"
]

# Keep only columns that exist in the dataset
feature_cols = [col for col in candidate_features if col in df_model.columns]

# Remove rows with missing values caused by lag and rolling features
df_model = df_model.dropna(subset=feature_cols + [target]).reset_index(drop=True)

print("Selected features:")
print(feature_cols)

print("\nModeling dataset shape:", df_model.shape)
```

```python
X = df_model[feature_cols]
y = df_model[target]

# Chronological split
split_index = int(len(df_model) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

time_test = df_model["time"].iloc[split_index:]

print("Training set shape:", X_train.shape)
print("Testing set shape:", X_test.shape)

print("Training period:", df_model["time"].iloc[0], "to", df_model["time"].iloc[split_index - 1])
print("Testing period:", df_model["time"].iloc[split_index], "to", df_model["time"].iloc[-1])
```

## 8. Forecast-Assisted Models

Three models are used to predict actual electricity demand. All models use the same forecast-assisted feature set, including `total load forecast` as an external predictor.

1. **Linear Regression**  
   This is the baseline model. It is simple and easy to explain.

2. **Random Forest Regressor**  
   This model can capture non-linear relationships between features and electricity demand.

3. **XGBoost Regressor**  
   This model is strong for structured tabular data and can handle complex feature interactions.

By comparing these three models, we show whether more advanced models improve actual demand prediction performance.


```python
models = {
    "Linear Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LinearRegression())
    ]),
    
    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ),
    
    "XGBoost": XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=5,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42
    )
}
```

```python
results = []

predictions = pd.DataFrame({
    "time": time_test.values,
    "actual": y_test.values
})

for model_name, model in models.items():
    print(f"Training {model_name}...")
    
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    
    results.append({
        "Model": model_name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse
    })
    
    predictions[model_name] = y_pred

results_df = pd.DataFrame(results).sort_values("RMSE").reset_index(drop=True)

display(results_df)
```

## 9. Model Evaluation

The models are evaluated using three metrics:

- **MAE**: average absolute prediction error
- **MSE**: average squared prediction error
- **RMSE**: square root of MSE, with the same unit as the target variable

A lower value means better performance when predicting `total load actual`.


```python
plt.figure(figsize=(9, 5))

plt.bar(
    results_df["Model"],
    results_df["RMSE"],
    color=COLOR_MODEL
)

plt.title("Model Comparison Based on RMSE")
plt.xlabel("Model")
plt.ylabel("RMSE")
plt.tight_layout()
plt.show()
```
